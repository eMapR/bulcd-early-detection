# Early Detection — interactive prototype

A working prototype of how an Early Detection application could feel: **choose an area → set the expectation and monitoring periods → run → view condition and timing → compare imagery → click a pixel to see why and when it was flagged.** Everything runs live in Google Earth Engine with the same BULC-D workflow as the demonstration notebooks (`early_detection_demo`, `north_cascades_demo`, `testsite_demo`); nothing about the method changes.

**This notebook is a functional prototype and specification for the future Early Detection interface, not the interface itself.** Because intended users may not be able to install local software, a Google Earth Engine (JavaScript) application is the current likely home for that interface; that isn't a final decision. Using this prototype should help settle:

- which controls users actually need;
- which parameters to expose, and which to keep advanced;
- how to present *condition* versus *timing*;
- how map and imagery comparison should work;
- how pixel inspection should work;

so the eventual interface translates a workflow that has been tried, rather than being designed blind.

## How to use it

Run the cell below once; everything else is buttons and sliders.

1. **Choose an area.** Draw a polygon or rectangle with the map's toolbar, pick a saved area, or enter an Earth Engine FeatureCollection asset ID (an example is offered). Press **Use this area**. The size is checked: over 100 km² gets a warning, over 1,000 km² is blocked unless allowed under *Advanced settings*.
2. **Set expectation and monitoring.**
   - **Expectation years** define what is *normal* for each pixel.
   - The **monitoring year** and **seasonal window** are what gets evaluated against that normal; the box under the sliders says this in words.
   - **Sensitivity** and **decision threshold** are the everyday controls.
   - Dampening, the posterior leveler, monitoring sensors and masks are under **Advanced settings**.
3. **Run Early Detection.** Results appear as switchable layers on the results map (top-right layer control): the condition map, the first-detected season, the NBR composite, and true color for three years. Summaries are in the tabs below the map.
4. **Click any pixel on the results map.** The *Pixel inspector* tab shows that location's observations against the expected seasonal value, its probability of decrease, the 0.5 threshold, when it first crossed, and every period above it.

The **Parameter behavior** tab runs the one-at-a-time sensitivity maps and response curves only when asked, because they take several BULC-D runs.

*Terms, as in the demonstration notebooks:* **condition** = below / within / above the expected state during the monitoring season (not necessarily when the change happened). **First detected** = the first season, and date, in which BULC-D's probability crossed the threshold; detection timing, not a verified disturbance date.

## What to watch for while trying it

- **Speed.** Everything is computed live. For a ~25 km² area, a run takes about 2 minutes, and each pixel inspection about 30 s per season. Larger areas are slower; a deployed interface would need precomputed products or smaller default areas.
- **Condition vs timing.** Are the two products, and their different questions, clear from the tabs and layers?
- **Controls.** Which sliders did you actually need? Which advanced settings did you touch?
- **Pixel inspector.** Does the chart explain *why* a pixel was flagged and *when*?
